```python
## Write a Python program to count the occurrences of each word in a given sentence.
# Program to count occurrences of each word in a sentence

sentence = input("Enter a sentence: ")

words = sentence.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("Word occurrences:")
for word, count in word_count.items():
    print(word, ":", count)
```

```python
## Write a Python program to add 'ing' at the end of a given string (length should beat least 3). If the given string already ends with 'ing' then add 'ly' instead If the string length of the given string is less than 3, leave it unchanged
string = input("Enter a string: ")

if len(string) < 3:
    result = string
elif string.endswith("ing"):
    result = string + "ly"
else:
    result = string + "ing"

print("Result:", result)
```

```python
## Write a Python program to find the first appearance of the substring 'not' and'poor' from a given string, if 'not' follows the 'poor', replace the whole 'not'...'poor'substring with 'good'. Return the resulting string

string = input("Enter a string: ")

not_pos = string.find("not")
poor_pos = string.find("poor")

if not_pos != -1 and poor_pos != -1 and not_pos < poor_pos:
    string = string[:not_pos] + "good" + string[poor_pos + 4:]

print("Result:", string)
```

```python
## Program to find Greatest Common Divisor of two numbers. For example, the GCDof20 and 28 is 4 and the GCD of 98 and 56 is 14.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD is:", a)
```

```python
## Write a Python program to check whether a list contains a sublist


main_list = [1, 2, 3, 4, 5, 6]
sub_list = [2, 3, 4]

found = False

for i in range(len(main_list) - len(sub_list) + 1):
    if main_list[i:i + len(sub_list)] == sub_list:
        found = True
        break

if found:
    print("Sublist found in the list")
else:
    print("Sublist not found in the list")
```

```python
## Write a Python program to sort a dictionary (ascending /descending) by value.


my_dict = {'a': 90, 'b': 70, 'c': 35, 'd': 20}
asc_dict = dict(sorted(my_dict.items(), key=lambda item: item[1]))
desc_dict = dict(sorted(my_dict.items(), key=lambda item: item[1], reverse=True))

print("Ascending order by value:", asc_dict)
print("Descending order by value:", desc_dict)
```

```python
## Given a number n, write a python program to make and print the list of Fibonacci series up to n. Input : n=7 Hint : first 7 numbers in the series Expected output : First few Fibonacci numbers are 0, 1, 1, 2, 3, 5, 8, 13

n = int(input("Enter n: "))

a, b = 0, 1
fibonacci = []

for i in range(n):
    fibonacci.append(a)
    a, b = b, a + b


print("First few Fibonacci numbers are:", ", ".join(map(str, fibonacci)))
```

```python
## Counting the frequencies in a list using a dictionary in Python. Input : [1, 1, 1, 5,5, 3, 1, 3, 3, 1,4, 4, 4, 2, 2, 2, 2]
numbers = [1, 1, 1, 5, 5, 3, 1, 3, 3, 1, 4, 4, 4, 2, 2, 2, 2]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print("Frequencies of elements:")
for key, value in frequency.items():
    print(key, ":", value)
```

```python
## Write a python program using function to find the sum of odd series and evenseries
## Odd series: 12/ 1! +32/ 3! + 52/ 5!+……n
## Even series: 22/ 2! + 42/ 4! + 62/ 6!+……n

import math


def odd_series_sum(n):
    sum_odd = 0
    for i in range(1, n+1, 2):
        sum_odd += (i**2) / math.factorial(i)
    return sum_odd


def even_series_sum(n):
    sum_even = 0
    for i in range(2, n+1, 2):
        sum_even += (i**2) / math.factorial(i)
    return sum_even

n = int(input("Enter n: "))

sum_odd = odd_series_sum(n)
sum_even = even_series_sum(n)

print("Sum of odd series:", sum_odd)
print("Sum of even series:", sum_even)
```

```python
## Python Program to Find Factorial of Number Using Recursion

def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = int(input("Enter a number: "))

result = factorial(num)
print(f"Factorial of {num} is {result}")
```

```python
## Write a Python function that takes a list and returns a new list with unique elements of the first list

def unique_list(lst):
    unique = []
    for item in lst:
        if item not in unique:
            unique.append(item)
    return unique


original_list = [1, 2, 3, 2,3, 4,5, 1, 5]
new_list = unique_list(original_list)

print("Original list:", original_list)
print("Unique list:", new_list)
```

```python
## refernce taken from website geeks from geeks

## Mini project :
#Problem Statement : Password Generator Make a program to generate a strong password using the input given by the user.To generate a password, randomly take some words from the user input and theninclude numbers, special characters and capital letters to generate the password.Also, keep a check that password length is more than 8 characters.Note: Include Exception handling wherever required. Also, make a ‘User’ class and store the details like user id, name and password of each user as a tuple.
import random
import string


class User:
    def __init__(self, user_id, name, password):
        self.user_id = user_id
        self.name = name
        self.password = password

    def get_details(self):
        return (self.user_id, self.name, self.password)


def generate_password(user_input):
    try:
        if not user_input.strip():
            raise ValueError("Input cannot be empty!")

        words = user_input.split()
        if len(words) == 0:
            raise ValueError("Please enter at least one word.")


        selected_words = random.sample(words, min(len(words), random.randint(2, 3)))
        password = ''.join(selected_words)

        password += str(random.randint(10, 99))


        password += random.choice(string.punctuation)

        password = ''.join(random.choice([c.upper(), c]) for c in password)


        while len(password) < 8:
            password += random.choice(string.ascii_letters + string.digits + string.punctuation)

        return password

    except ValueError as ve:
        print("Error:", ve)
        return None
    except Exception as e:
        print("An unexpected error occurred:", e)
        return None


users = []

try:
    num_users = int(input("Enter number of users to create: "))
    for i in range(num_users):
        print(f"\nUser {i+1}:")
        user_id = input("Enter user ID: ")
        name = input("Enter name: ")
        user_input = input("Enter words for password generation: ")

        password = generate_password(user_input)
        if password:
            user = User(user_id, name, password)
            users.append(user)
            print("Generated password:", password)
        else:
            print("Password generation failed.")

except ValueError:
    print("Please enter a valid number for users.")
except Exception as e:
    print("An unexpected error occurred:", e)

print("\nAll User Details (user_id, name, password):")
for user in users:
    print(user.get_details())
```
